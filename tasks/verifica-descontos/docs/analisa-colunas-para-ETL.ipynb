## Unir `DF_CONTRATACOES_COLUNAS` e `DF_ITEM_HOMOLOGADO_COLUNAS` com seus respectivos dados legados vindos do script `tasks\verifica-descontos\docs\analisa-descontos-dados-legados.ipynb`, de forma que os datasets atuais do banco recebam as colunas necessárias para as informações de desontos

### `DF_ITEM_HOMOLOGADO_COLUNAS`

```python
# DF_ITEM_HOMOLOGADO_COM_LEGADOS = DF_ITEM_HOMOLOGADO_COLUNAS.merge(
#     DF_LEGADOS_ITEM_HOMOLOGADO,
#     on=["numero_controle_pncp", "numero_item"],
#     how="left",
#     suffixes=("", "_legado"),
#     indicator=True
# )

# DF_ITEM_HOMOLOGADO_COM_LEGADOS["_merge"].value_counts()
# linhas_com_legado = DF_ITEM_HOMOLOGADO_COM_LEGADOS.query('_merge == "both"')
# linhas_sem_legado = DF_ITEM_HOMOLOGADO_COM_LEGADOS.query('_merge == "left_only"')
# linhas_com_legado.head()
# linhas_sem_legado.head()

#-----------------------------------------------------------------------
#  MANEIRA CORRETA DE FAZER O MERGE COM O LEGADO, POIS ALGUMAS CHAVES SE REPETEM NO LEGADO.
#-----------------------------------------------------------------------
# A correspondência entre os datasets é feita no nível do item.
chaves_item = ["numero_controle_pncp", "numero_item"]
colunas_legado = [
    coluna
    for coluna in DF_LEGADOS_ITEM_HOMOLOGADO.columns
    if coluna not in chaves_item
]

# Algumas chaves se repetem no legado. Um valor único permanece escalar; valores
# diferentes são preservados em lista para evitar perda de informação.
def preservar_valores_unicos(serie):
    valores = serie.dropna().unique().tolist()
    if not valores:
        return pd.NA
    return valores[0] if len(valores) == 1 else valores


chaves_repetidas = DF_LEGADOS_ITEM_HOMOLOGADO.duplicated(
    subset=chaves_item, keep=False
)
legados_sem_repeticao = DF_LEGADOS_ITEM_HOMOLOGADO.loc[~chaves_repetidas]
legados_repetidos_agrupados = (
    DF_LEGADOS_ITEM_HOMOLOGADO.loc[chaves_repetidas]
    .groupby(chaves_item, as_index=False, dropna=False)[colunas_legado]
    .agg(preservar_valores_unicos)
)
# Reúne verticalmente as linhas que já eram únicas e as repetições consolidadas.
DF_LEGADOS_ITEM_HOMOLOGADO_AGRUPADO = pd.concat(
    # Estes dois DataFrames têm as mesmas colunas e serão empilhados.
    [legados_sem_repeticao, legados_repetidos_agrupados],
    # Recria o índice para que ele seja contínuo após a concatenação.
    ignore_index=True
)

# Enriquece os itens atuais com as colunas do legado já consolidado.
DF_ITEM_HOMOLOGADO_COM_LEGADOS = DF_ITEM_HOMOLOGADO_COLUNAS.merge(
    DF_LEGADOS_ITEM_HOMOLOGADO_AGRUPADO, # DataFrame da direita, com no máximo uma linha por chave de item.
    on=chaves_item,  # Faz a correspondência pela contratação e pelo número do item.
    how="left", # Preserva todas as linhas do dataset atual, mesmo sem correspondência.
    suffixes=("", "_legado"), # Mantém nomes atuais e acrescenta '_legado' somente em colunas repetidas.
    indicator=True,  # Cria '_merge' para indicar 'both', 'left_only' ou 'right_only'.
    validate="many_to_one" # Garante várias linhas à esquerda e no máximo uma por chave à direita.
)

assert len(DF_ITEM_HOMOLOGADO_COM_LEGADOS) == len(DF_ITEM_HOMOLOGADO_COLUNAS), (
    "A junção alterou a quantidade de linhas de DF_ITEM_HOMOLOGADO_COLUNAS"
)

print(f"O dataset após o merge contém {DF_ITEM_HOMOLOGADO_COM_LEGADOS.shape[0]} linhas")

print(f"O dataset atual do BANCO DE DADOS contém {DF_ITEM_HOMOLOGADO_COLUNAS.shape[0]} linhas")
print(f"Diferença de {DF_ITEM_HOMOLOGADO_COM_LEGADOS.shape[0] - DF_ITEM_HOMOLOGADO_COLUNAS.shape[0]} linhas")
DF_ITEM_HOMOLOGADO_COM_LEGADOS.head()
```

```python
# Linhas que receberam informações do legado
linhas_com_legado = DF_ITEM_HOMOLOGADO_COM_LEGADOS.query('_merge == "both"')

print(f"O dataset após merge contém {DF_ITEM_HOMOLOGADO_COM_LEGADOS.shape[0]} linhas")
print(f"Linhas com informações do legado: {linhas_com_legado.shape[0]}")
linhas_com_legado.head()
```

```python
# Linhas que não receberam informações do legado
linhas_sem_legado = DF_ITEM_HOMOLOGADO_COM_LEGADOS.query('_merge == "left_only"')

print(f"Linhas que não tiveram correspondência no legado: {linhas_sem_legado.shape[0]}")
linhas_sem_legado.head()
```

```python
total = len(DF_ITEM_HOMOLOGADO_COM_LEGADOS)

print(f"Total de linhas atuais: {total}")
print(f"Linhas impactadas por dados legados: {linhas_com_legado.shape[0]}")
print(f"Linhas sem dados legados: {linhas_sem_legado.shape[0]}")
print(f"Cobertura: {linhas_com_legado.shape[0] / total:.2%}")
```

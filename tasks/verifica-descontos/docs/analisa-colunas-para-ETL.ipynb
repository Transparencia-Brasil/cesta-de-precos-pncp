## Unir `DF_CONTRATACOES_COLUNAS` e `DF_ITEM_HOMOLOGADO_COLUNAS` com seus respectivos dados legados vindos do script `tasks\verifica-descontos\docs\analisa-descontos-dados-legados.ipynb`, de forma que os datasets atuais do banco recebam as colunas necessárias para as informações de desontos

### `DF_ITEM_HOMOLOGADO_COLUNAS`

```python
# A correspondência entre os datasets é feita no nível do item.
chaves_item = ["numero_controle_pncp", "numero_item"]
colunas_legado = [
    coluna
    for coluna in DF_LEGADOS_ITEM_HOMOLOGADO.columns
    if coluna not in chaves_item
]

# Algumas chaves se repetem no legado. Um valor único permanece escalar; valores
# diferentes são preservados em lista para evitar perda de informação.
def preservar_valores_unicos(serie):
    valores = serie.dropna().unique().tolist()
    if not valores:
        return pd.NA
    return valores[0] if len(valores) == 1 else valores


chaves_repetidas = DF_LEGADOS_ITEM_HOMOLOGADO.duplicated(
    subset=chaves_item, keep=False
)
legados_sem_repeticao = DF_LEGADOS_ITEM_HOMOLOGADO.loc[~chaves_repetidas]
legados_repetidos_agrupados = (
    DF_LEGADOS_ITEM_HOMOLOGADO.loc[chaves_repetidas]
    .groupby(chaves_item, as_index=False, dropna=False)[colunas_legado]
    .agg(preservar_valores_unicos)
)
# Reúne verticalmente as linhas que já eram únicas e as repetições consolidadas.
DF_LEGADOS_ITEM_HOMOLOGADO_AGRUPADO = pd.concat(
    # Estes dois DataFrames têm as mesmas colunas e serão empilhados.
    [legados_sem_repeticao, legados_repetidos_agrupados],
    # Recria o índice para que ele seja contínuo após a concatenação.
    ignore_index=True
)

# Enriquece os itens atuais com as colunas do legado já consolidado.
DF_ITEM_HOMOLOGADO_COM_LEGADOS = DF_ITEM_HOMOLOGADO_COLUNAS.merge(
    DF_LEGADOS_ITEM_HOMOLOGADO_AGRUPADO, # DataFrame da direita, com no máximo uma linha por chave de item.
    on=chaves_item,  # Faz a correspondência pela contratação e pelo número do item.
    how="left", # Preserva todas as linhas do dataset atual, mesmo sem correspondência.
    suffixes=("", "_legado"), # Mantém nomes atuais e acrescenta '_legado' somente em colunas repetidas.
    indicator=True,  # Cria '_merge' para indicar 'both', 'left_only' ou 'right_only'.
    validate="many_to_one" # Garante várias linhas à esquerda e no máximo uma por chave à direita.
)

assert len(DF_ITEM_HOMOLOGADO_COM_LEGADOS) == len(DF_ITEM_HOMOLOGADO_COLUNAS), (
    "A junção alterou a quantidade de linhas de DF_ITEM_HOMOLOGADO_COLUNAS"
)

print(f"O dataset após o merge contém {DF_ITEM_HOMOLOGADO_COM_LEGADOS.shape[0]} linhas")

print(f"O dataset atual do BANCO DE DADOS contém {DF_ITEM_HOMOLOGADO_COLUNAS.shape[0]} linhas")
print(f"Diferença de {DF_ITEM_HOMOLOGADO_COM_LEGADOS.shape[0] - DF_ITEM_HOMOLOGADO_COLUNAS.shape[0]} linhas")
DF_ITEM_HOMOLOGADO_COM_LEGADOS.head()
```

```python
# Linhas que receberam informações do legado
linhas_com_legado = DF_ITEM_HOMOLOGADO_COM_LEGADOS.query('_merge == "both"')

print(f"O dataset após merge contém {DF_ITEM_HOMOLOGADO_COM_LEGADOS.shape[0]} linhas")
print(f"Linhas com informações do legado: {linhas_com_legado.shape[0]}")
linhas_com_legado.head()
```

```python
# Linhas que não receberam informações do legado
linhas_sem_legado = DF_ITEM_HOMOLOGADO_COM_LEGADOS.query('_merge == "left_only"')

print(f"Linhas que não tiveram correspondência no legado: {linhas_sem_legado.shape[0]}")
linhas_sem_legado.head()
```

#### Dados e métricas pós análise:

```python
total = len(DF_ITEM_HOMOLOGADO_COM_LEGADOS)

print(f"Total de linhas atuais: {total}")
print(f"Linhas impactadas por dados legados: {linhas_com_legado.shape[0]}")
print(f"Linhas sem dados legados: {linhas_sem_legado.shape[0]}")
print(f"Cobertura: {linhas_com_legado.shape[0] / total:.2%}")
```

### `DF_CONTRATACOES_COLUNAS`

```python
chave_contratacao = ["numero_controle_pncp"]
colunas_legado_contratacoes = [
    coluna
    for coluna in DF_LEGADOS_CONTRATACOES.columns
    if coluna not in chave_contratacao
]

chaves_contratacoes_repetidas = DF_LEGADOS_CONTRATACOES.duplicated(
    subset=chave_contratacao, keep=False
)
contratacoes_legadas_sem_repeticao = DF_LEGADOS_CONTRATACOES.loc[
    ~chaves_contratacoes_repetidas
]
contratacoes_legadas_repetidas_agrupadas = (
    DF_LEGADOS_CONTRATACOES.loc[chaves_contratacoes_repetidas]
    .groupby(
        chave_contratacao, as_index=False, dropna=False
    )[colunas_legado_contratacoes]
    .agg(preservar_valores_unicos)
)
# Reúne verticalmente as contratações únicas e as repetições consolidadas.
DF_LEGADOS_CONTRATACOES_AGRUPADO = pd.concat(
    # Os DataFrames da lista serão empilhados, preservando as mesmas colunas.
    [
        # Contratações que já possuíam uma única linha no legado.
        contratacoes_legadas_sem_repeticao,
        # Contratações repetidas após a preservação dos valores únicos.
        contratacoes_legadas_repetidas_agrupadas,
    ],
    # Descarta os índices anteriores e cria um índice contínuo.
    ignore_index=True
)

# Enriquece as contratações atuais com as colunas do legado consolidado.
DF_CONTRATACOES_COM_LEGADOS = DF_CONTRATACOES_COLUNAS.merge(
    # DataFrame da direita, com uma única linha para cada contratação.
    DF_LEGADOS_CONTRATACOES_AGRUPADO,
    # Relaciona os datasets pelo identificador único da contratação no PNCP.
    on=chave_contratacao,
    # Preserva todas as contratações atuais, mesmo sem dados legados.
    how="left",
    # Mantém nomes atuais e acrescenta '_legado' em eventuais colisões.
    suffixes=("", "_legado"),
    # Cria '_merge' para permitir a auditoria das correspondências.
    indicator=True,
    # Confirma que existe no máximo uma linha por chave em cada lado.
    validate="one_to_one"
)

assert len(DF_CONTRATACOES_COM_LEGADOS) == len(DF_CONTRATACOES_COLUNAS), (
    "A junção alterou a quantidade de linhas de DF_CONTRATACOES_COLUNAS"
)

print(f"O dataset após o merge contém {DF_CONTRATACOES_COM_LEGADOS.shape[0]} linhas")
print(f"O dataset atual do BANCO DE DADOS contém {DF_CONTRATACOES_COLUNAS.shape[0]} linhas")
print(f"Diferença de {DF_CONTRATACOES_COM_LEGADOS.shape[0] - DF_CONTRATACOES_COLUNAS.shape[0]} linhas")
DF_CONTRATACOES_COM_LEGADOS.head()
```

```python
# Linhas que receberam informações do legado
linhas_contratacoes_com_legado = DF_CONTRATACOES_COM_LEGADOS.query('_merge == "both"')

print(f"Linhas com informações do legado: {linhas_contratacoes_com_legado.shape[0]}")
linhas_contratacoes_com_legado.head()
```

```python
# Linhas que não receberam informações do legado
linhas_contratacoes_sem_legado = DF_CONTRATACOES_COM_LEGADOS.query('_merge == "left_only"')

print(f"Linhas que não tiveram correspondência no legado: {linhas_contratacoes_sem_legado.shape[0]}")
linhas_contratacoes_sem_legado.head()
```

#### Dados e métricas pós análise:

```python
total_contratacoes = len(DF_CONTRATACOES_COM_LEGADOS)
impactadas_contratacoes = linhas_contratacoes_com_legado.shape[0]
sem_legado_contratacoes = linhas_contratacoes_sem_legado.shape[0]

print(f"Total de linhas atuais: {total_contratacoes}")
print(f"Linhas impactadas por dados legados: {impactadas_contratacoes}")
print(f"Linhas sem dados legados: {sem_legado_contratacoes}")
print(f"Cobertura: {impactadas_contratacoes / total_contratacoes:.2%}")
```

## Integrar `srp` das contratações aos itens homologados

A coluna `srp` é associada aos itens por `numero_controle_pncp`. A junção é do tipo `left`, portanto mantém todas as linhas de `DF_ITEM_HOMOLOGADO_COM_LEGADOS` e preenche `srp` apenas quando a contratação correspondente existe.

```python
quantidade_itens_antes_srp = len(DF_ITEM_HOMOLOGADO_COM_LEGADOS)

# Seleciona somente a chave da contratação e a coluna que será levada aos itens.
# Cria uma cópia independente para evitar alterações acidentais no DataFrame original.
DF_SRP_POR_CONTRATACAO = DF_CONTRATACOES_COM_LEGADOS[
    ["numero_controle_pncp", "srp"]
].copy()

# Associa o indicador de SRP de cada contratação aos seus respectivos itens.
DF_ITEM_HOMOLOGADO_COM_LEGADOS_E_SRP = DF_ITEM_HOMOLOGADO_COM_LEGADOS.merge(
    DF_SRP_POR_CONTRATACAO, # DataFrame da direita, contendo uma linha por contratação e seu valor de srp
    on="numero_controle_pncp", # Repete o valor de srp nos itens que possuem o mesmo número de controle PNCP.
    how="left", # Mantém todos os itens, inclusive quando a contratação não for encontrada.
    validate="many_to_one", # Valida que vários itens podem apontar para somente uma contratação.
    indicator="_merge_srp" # Cria uma coluna específica para auditar a correspondência desta junção.
)

assert len(DF_ITEM_HOMOLOGADO_COM_LEGADOS_E_SRP) == quantidade_itens_antes_srp, (
    "A integração de srp alterou a quantidade de itens homologados"
)

itens_com_srp = DF_ITEM_HOMOLOGADO_COM_LEGADOS_E_SRP["_merge_srp"].eq("both").sum()
itens_sem_contratacao = (
    DF_ITEM_HOMOLOGADO_COM_LEGADOS_E_SRP["_merge_srp"].eq("left_only").sum()
)

print(f"Itens antes da integração de srp: {quantidade_itens_antes_srp}")
print(f"Itens após a integração de srp: {len(DF_ITEM_HOMOLOGADO_COM_LEGADOS_E_SRP)}")
print(f"Itens associados a uma contratação: {itens_com_srp}")
print(f"Itens sem contratação correspondente: {itens_sem_contratacao}")
DF_ITEM_HOMOLOGADO_COM_LEGADOS_E_SRP.head()
```

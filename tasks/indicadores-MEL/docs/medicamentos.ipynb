```python
import pyreadr
from pathlib import Path

# Localiza a raiz do repositório pelo marcador AGENTS.md
def _repo_root():
    p = Path().resolve()
    for d in [p] + list(p.parents):
        if (d / "AGENTS.md").exists():
            return d
    return p

rds_path = _repo_root() / "tasks" / "indicadores-MEL" / "inputs" / "medicamentos.rds"
df = pyreadr.read_r(rds_path)[None]
```
